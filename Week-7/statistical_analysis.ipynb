# Week 7: Statistical Business Analysis

This notebook performs descriptive statistics, distribution analysis, hypothesis testing, confidence intervals, correlation analysis, and regression analysis.

```
import pandas as pd
from scipy import stats
import statsmodels.api as sm

business_data = pd.read_csv('business_data.csv')
business_data.head()
```

```
numeric = ['Quantity', 'Price', 'Marketing_Spend', 'Revenue', 'Profit', 'Tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
descriptive = business_data[numeric].agg(['count', 'mean', 'median', 'std', 'min', 'max']).T
descriptive['mode'] = business_data[numeric].mode().iloc[0]
descriptive
```

```
correlation = business_data[numeric].corr()
correlation
```

```
high = business_data[business_data['Marketing_Spend'] >= business_data['Marketing_Spend'].median()]['Revenue']
low = business_data[business_data['Marketing_Spend'] < business_data['Marketing_Spend'].median()]['Revenue']
stats.ttest_ind(high, low, equal_var=False)
```

```
revenue_mean = business_data['Revenue'].mean()
revenue_ci = stats.t.interval(0.95, len(business_data)-1, loc=revenue_mean, scale=stats.sem(business_data['Revenue']))
revenue_mean, revenue_ci
```

```
X = sm.add_constant(business_data[['Marketing_Spend', 'Quantity', 'Price', 'Tenure', 'MonthlyCharges']])
model = sm.OLS(business_data['Revenue'], X).fit()
model.rsquared, model.summary()
```
