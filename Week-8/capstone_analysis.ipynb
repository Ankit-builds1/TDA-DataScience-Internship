End-to-end business analysis workflow.

```
import pandas as pd
from scipy import stats
import statsmodels.api as sm

churn = pd.read_csv('data/cleaned_data.csv')
churn.head()
```

```
churn.describe(include='all')
```

```
churn.groupby('Contract')['Churn'].agg(['mean','count']).sort_values('mean', ascending=False)
```

```
stats.ttest_ind(churn[churn.Churn==1]['Tenure'], churn[churn.Churn==0]['Tenure'], equal_var=False)
```

```
pd.crosstab(churn['Contract'], churn['Churn'])
```

```
encoded = pd.get_dummies(churn[['Tenure','MonthlyCharges','TotalCharges','SeniorCitizen','Contract','PaymentMethod','PaperlessBilling']], drop_first=True).astype(float)
model = sm.Logit(churn['Churn'], sm.add_constant(encoded)).fit(disp=False)
model.summary()
```

## Recommendation

Use predicted risk and customer value to create an urgent-save retention queue.
