# Week 5: Customer Sales Analysis

This notebook performs advanced Pandas manipulation, filtering, aggregation, pivot-table summarization, merge auditing, and business insight generation.

```python

import pandas as pd
from analysis import (
    advanced_filters,
    clean_churn_data,
    clean_sales_data,
    compute_sales_summary,
    format_inr,
    load_churn_data,
    load_sales_data,
)

sales = clean_sales_data(load_sales_data("sales_data.csv"))
churn = clean_churn_data(load_churn_data("customer_data.csv"))
filters = advanced_filters(sales)
summary = compute_sales_summary(sales, churn)

print("CUSTOMER SALES ANALYSIS REPORT")
print("Total Revenue:", format_inr(summary["total_revenue"]))
print("Total Customers:", summary["total_customers"])
print("Average Order Value:", format_inr(summary["average_order_value"]))
print("Top Customer:", summary["top_customer"], "-", format_inr(summary["top_customer_revenue"]))
print("Top Product:", summary["top_product"], "-", format_inr(summary["top_product_revenue"]))
print("Top Region:", summary["top_region"], "-", format_inr(summary["top_region_revenue"]))
print("Overall Churn Rate:", f"{summary['overall_churn_rate'] * 100:.2f}%")
```
